# 📊 Reto 3 – Funciones y estadísticas en Pandas

**Objetivo:** aplicar funciones estadísticas y crear nuevas columnas basadas en cálculos.

```python
import pandas as pd

datos = pd.read_csv('4_resultados_agua.csv')

datos['categoria_ph'] = pd.cut(
    datos['pH'],
    bins=[0, 6.5, 8.5, 14],
    labels=['Ácido', 'Neutro', 'Básico']
)

datos.head()
```

```python
datos.mean(numeric_only=True)
```

```python
datos.groupby('categoria_ph').mean(numeric_only=True)
```

```python
datos[datos['categoria_ph'] == 'Ácido']
```

```python
datos['ph_normalizado'] = (datos['pH'] - datos['pH'].mean()) / datos['pH'].std()
```
