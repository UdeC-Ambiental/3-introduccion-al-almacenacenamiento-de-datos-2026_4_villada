```python
import pandas as pd

datos = pd.read_csv('4_resultados_agua.csv')

datos['categoria_ph'] = pd.cut(
    datos['pH'],
    bins=[0, 6.5, 8.5, 14],
    labels=['Ácido', 'Neutro', 'Básico']
)

datos.head()
```

```python
datos.mean(numeric_only=True)
```

```python
datos.groupby('categoria_ph').mean(numeric_only=True)
```

```python
agua_acida = datos[datos['categoria_ph'] == 'Ácido']
agua_neutra = datos[datos['categoria_ph'] == 'Neutro']
agua_basica = datos[datos['categoria_ph'] == 'Básico']

print("Muestras ácidas:")
print(agua_acida.head())

print("\nMuestras neutras:")
print(agua_neutra.head())

print("\nMuestras básicas:")
print(agua_basica.head())
```

```python
resumen = datos.select_dtypes(include='number').groupby(datos['categoria_ph']).agg(['mean', 'min', 'max'])

print(resumen)
```

Se realizó un filtrado independiente para cada categoría de pH (ácido, neutro y básico), con el fin de analizar sus características específicas. Posteriormente, se generó un resumen estadístico agrupado, obteniendo valores promedio, mínimos y máximos para cada categoría.
