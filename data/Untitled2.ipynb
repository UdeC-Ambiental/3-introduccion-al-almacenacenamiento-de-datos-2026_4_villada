```python
import pandas as pd

df = pd.read_csv('4_resultados_agua.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
df['estado_pH'] = pd.cut(df['pH'], bins=[0,6.5,8.5,14], labels=['Ácido','Normal','Alcalino'])
df['estado_temp'] = pd.cut(df['Temp'], bins=[0,20,30,100], labels=['Baja','Normal','Alta'])
df['OD_promedio'] = pd.cut(df['OD_promedio'], bins=[0,5,8,20], labels=['Bajo','Aceptable','Óptimo'])
```

```python
df['diagnostico'] = df['estado_ph'].astype(str) + ' - ' + \
                    df['estado_temp'].astype(str) + ' - ' + \
                    df['OD_promedio'].astype(str)
```

```python
resumen = df[['pH', 'estado_temp', 'OD_promedio']].describe()
resumen
```

```python
df['estado_ph'].value_counts()
df['estado_temp'].value_counts()
df['OD_promedio'].value_counts()
```

```python
diagnostico_final = df.groupby('diagnostico').size().reset_index(name='conteo')
diagnostico_final
```

```python
conclusion = diagnostico_final.sort_values(by='conteo', ascending=False)
conclusion.head(1)
```

```python
df.to_csv('resultado_final_analisis.csv', index=False)
```

El respectivo análisis del parametro de pH, la temperatura y el oxígeno disuelto muestran una cierta condición predominante del agua el cual es el resultado más frecuente, lo que indica que la calidad del agua es tanto buena/moderada/crítica según los valores arrojados.
